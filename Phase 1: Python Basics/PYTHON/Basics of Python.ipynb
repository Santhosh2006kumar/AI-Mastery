**Comprehensions**

```python
#squared numbers
list2 = [1,2,3]
squared_list2 = [i**2 for i in list2]
print(squared_list2)
```

```python
#only odd numbers
odd_list2 = [i for i in list2 if i%2!=0]
print(odd_list2)
```

```python
#squared dict
sq_dict = {n : n for n in list2}
print(sq_dict)
```
